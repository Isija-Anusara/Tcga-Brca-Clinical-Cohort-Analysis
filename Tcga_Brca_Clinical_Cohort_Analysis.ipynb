```python

import csv

Full_Record_Of_The_Patients = []

with open("data_clinical_patient.txt" , "r" , encoding="utf-8") as file:
      reader = csv.DictReader(file, delimiter="\t")
      for row in reader:
                        Patient_ID = row["PATIENT_ID"]
                        Overall_Survival_Status = row["OS_STATUS"]
                        Age = row["AGE"]
                        if Age not in ("" , "[Not Available]"):
                                                               Age_Float = float(Age)
                                                               A_Locked_Tuple = (Patient_ID , Overall_Survival_Status , Age_Float )
                                                               Full_Record_Of_The_Patients.append(A_Locked_Tuple)
living_ages = []
deseased_ages = []
for Patient_ID, Overall_Survival_Status , Age_Float in Full_Record_Of_The_Patients:
  if Overall_Survival_Status == "0:LIVING":
    living_ages.append(Age_Float)
  elif Overall_Survival_Status == "1:DECEASED":
    deseased_ages.append(Age_Float)
if  sum(living_ages)/ len(living_ages) > sum(deseased_ages)/ len(deseased_ages) :
  print(" It is obvious that in this TCGA-BRCA cohort, living patients had a higher average diagnosis age than deceased patients")
else:
  print(" It is obvious that in this in this TCGA-BRCA cohort, living patients had a lower average diagnosis age than deceased patients")





```
